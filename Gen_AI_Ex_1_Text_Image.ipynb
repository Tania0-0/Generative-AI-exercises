```python
from diffusers import StableDiffusionPipeline
import torch
from PIL import Image
import matplotlib.pyplot as plt

#load model
model_id="runwayml/stable-diffusion-v1-5"

pipe=StableDiffusionPipeline.from_pretrained(model_id,torch_dtype=torch.float16)

#Use GPU if available
device = "cuda" if torch.cuda.is_available() else "cpu"
pipe = pipe.to(device)

#Get prompt
prompt=input("Enter the prompt:")

#Generate image
image=pipe(prompt).images[0]

#save img
image.save("generated_image.png")

#Display img
plt.imshow(image)
plt.axis("off")
plt.show()

print("image saved!!")
```

```python
# 1. Import
from diffusers import StableDiffusionPipeline
import torch
import matplotlib.pyplot as plt

# 2. Load model
model = StableDiffusionPipeline.from_pretrained(
    "runwayml/stable-diffusion-v1-5",
    torch_dtype=torch.float16
)

# 3. Select device
device = "cuda" if torch.cuda.is_available() else "cpu"
model.to(device)

# 4. Get prompt
prompt = input("Enter prompt: ")

# 5. Generate
image = model(prompt).images[0]

# 6. Save and display
image.save("generated_image.png")
plt.imshow(image)
plt.axis("off")
plt.show()
```

```python
# 4. Get prompt
prompt = input("Enter prompt: ")

# 5. Generate
image = model(prompt, num_images_per_prompt=3).images

# 6. Save and display
image[0].save("generated_image.png")
plt.imshow(image[0])
plt.axis("off")
plt.show()

image[1].save("generated_image.png")
plt.imshow(image[1])
plt.axis("off")
plt.show()
```

```python
# 1. Import

from diffusers import StableDiffusionPipeline
import torch
import matplotlib.pyplot as plt

# 2. Load model
model = StableDiffusionPipeline.from_pretrained(
    "runwayml/stable-diffusion-v1-5",
    torch_dtype=torch.float16
)

# 3. Select device
device = "cuda" if torch.cuda.is_available() else "cpu"
model.to(device)

#prompt = input("Enter prompt: ")

# 5. Generate
def genimg(prompt):
  image = model(prompt).images[0]
  return image

# 6. Save and display
```

```python
interface=gr.Interface(
    fn=genimg,
    inputs=gr.Textbox(),
    outputs=gr.Image()
)

interface.launch(share=True)
```
