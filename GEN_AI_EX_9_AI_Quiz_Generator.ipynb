FLAN-T5 is the open-source AI model.
The user enters a topic and number of questions. The topic is converted into a prompt, which is sent to FLAN-T5. The generated text is then displayed by Gradio.

```python
from transformers import AutoTokenizer, AutoModelForCausalLM
import gradio as gr
import torch
```

```python
model_name = "Qwen/Qwen2.5-0.5B-Instruct"

tokenizer = AutoTokenizer.from_pretrained(model_name)

model = AutoModelForCausalLM.from_pretrained(
    model_name,
    torch_dtype="auto",
    device_map="auto"
)

print("Model loaded successfully!")
```

```python
def generate_one_question(topic):

    messages = [
        {
            "role": "system",
            "content": "You are a quiz generator. Create simple college-level multiple choice questions."
        },
        {
            "role": "user",
            "content": f"""
Create ONE multiple choice question about {topic}.

Use exactly this format:

Question: [question]
A. [option]
B. [option]
C. [option]
D. [option]
Answer: [A/B/C/D]

Do not add explanations.
"""
        }
    ]

    text = tokenizer.apply_chat_template(
        messages,
        tokenize=False,
        add_generation_prompt=True
    )

    inputs = tokenizer(
        text,
        return_tensors="pt"
    ).to(model.device)

    outputs = model.generate(
        **inputs,
        max_new_tokens=150,
        do_sample=True,
        temperature=0.7
    )

    new_tokens = outputs[0][inputs["input_ids"].shape[1]:]

    result = tokenizer.decode(
        new_tokens,
        skip_special_tokens=True
    )

    return result.strip()
```

```python
def generate_quiz(topic, number):

    number = int(number)

    quiz = ""

    for i in range(number):

        result = generate_one_question(topic)

        quiz += f"\nQuestion {i+1}\n"
        quiz += result
        quiz += "\n"
        quiz += "-" * 50
        quiz += "\n"

    return quiz
```

```python
print(generate_quiz("Machine Learning", 3))
```

```python
interface = gr.Interface(
    fn=generate_quiz,
    inputs=[
        gr.Textbox(
            label="Enter Topic",
            placeholder="Example: Machine Learning"
        ),
        gr.Number(
            label="Number of Questions",
            value=5
        )
    ],
    outputs=gr.Textbox(
        label="Generated Quiz",
        lines=30
    ),
    title="AI Quiz Generator",
    description="Generate multiple choice questions using an open-source AI model."
)

interface.launch()
```

```python
from transformers import pipeline
import gradio as gr

# Load Qwen model
generator = pipeline(
    "text-generation",
    model="Qwen/Qwen2.5-0.5B-Instruct"
)

print("Model loaded successfully!")


```

```python
def generate_question(topic):

    prompt = f"""You are an MCQ generator.

Topic: {topic}

Write exactly:

Question: one question
A. option
B. option
C. option
D. option
Answer: A

Only write the MCQ. No explanation. No code."""

    result = generator(
        prompt,
        max_new_tokens=100,
        max_length=None,
        do_sample=True,
        temperature=0.3,
        return_full_text=False
    )

    return result[0]["generated_text"].strip()
```

```python
# Generate multiple questions
def generate_quiz(topic, number):

    quiz = ""

    for i in range(int(number)):

        question = generate_question(topic)

        quiz += f"\nQuestion {i+1}\n"
        quiz += question
        quiz += "\n"
        quiz += "-" * 50
        quiz += "\n"

    return quiz

```

```python

# Test
print(generate_quiz("Deep Learning", 3))
```

```python


# Simple Gradio interface
interface = gr.Interface(
    fn=generate_quiz,
    inputs=[
        gr.Textbox(label="Enter Topic"),
        gr.Number(label="Number of Questions", value=5)
    ],
    outputs=gr.Textbox(label="Generated Quiz", lines=30),
    title="AI Quiz Generator"
)

interface.launch()
```
